```python
grupa_a = {'python', 'pandas', 'numpy', 'sql', 'tableau'}
grupa_b = {'python', 'r', 'sql', 'ggplot2', 'dplyr'}
```

```python
suma1 = grupa_a | grupa_b
```

```python
wspolne1 = grupa_a & grupa_b
```

```python
roznica = grupa_a - grupa_b
```
