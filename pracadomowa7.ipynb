Zadanie 9

```python
wagi = [70, 85, 62, 95, 78]
wzrosty = [1.75, 1.80, 1.65, 1.78, 1.72]
```

```python
for i in range(len(wagi)):
  bmi = wagi[i] / wzrosty[i] ** 2
  if bmi < 18.5:
    klasyfikacja = "Niedowaga"
  elif bmi >= 18.5 and bmi <= 24.9:
    klasyfikacja = "Norma"
  elif bmi >= 25.0 and bmi <= 29.9:
    klasyfikacja = "Nadwaga"
  else:
    klasyfikacja = "Otyłość"


  print(f"Pacjent {i + 1}: waga {wagi[i]} kg, wzrost {wzrosty[i]} m, BMI {bmi:.1f} - {klasyfikacja}")
```

Zadanie 15

```python
wyniki = [78, 82, 85, 88, 90, 92, 95, 97, 99, 45, 101, 103, 105, 180]
```

```python
posort = sorted(wyniki)
```

```python
Q1 = posort[len(posort) // 4]
```

```python
Q3 = posort[3 * len(posort) // 4]
```

```python
IQR = Q3 - Q1
```

```python
granice = [Q1 - 1.5 * IQR, Q3 + 1.5 * IQR]
```

```python
gorna = granice[0]
```

```python
dolna = granice[1]
```

```python
outliers = [x for x in wyniki if x < dolna or x > gorna]
```

```python
print("Q1:", Q1)
```

```python
print("Q3:", Q3)
```

```python
print("IQR:", IQR)
```

```python
print("Granice:", granice)
```

```python
print("outlier:", outliers)
```
