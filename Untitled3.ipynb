```python
waga = 70
```

```python
wzrost = 1.75
```

```python
BMI = waga / (wzrost ** 2 )
```

```python
BMI
```

```python
zaokrąglone = round(BMI, 2)
```

```python
print(f"BMI zaokrąglone: {zaokrąglone}")
```

```python
if BMI <= 18.5:
    print("Niedowaga")
if BMI > 18.5 and BMI < 25:
    print("Norma")
if BMI >= 25:
    print("Nadwaga")
```
