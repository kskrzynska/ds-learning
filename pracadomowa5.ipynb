Zadanie 9

```python
cm = [
[80, 10], # TN, FP
[5, 45] # FN, TP
]
```

```python
tn = cm[0][0]
fp = cm[0][1]
fn = cm[1][0]
tp = cm[1][1]
```

```python
print(f"TN = {tn}, FP = {fp}, FN = {fn}, TP = {tp}")
```

```python
accuracy = (tp + tn) / (tp + tn + fp + fn)
```

```python
print(accuracy)
```

```python

precision = tp / (tp + fp)
```

```python
print(precision)
```

```python
recall = tp / (tp + fn)
```

```python
print(recall)
```

```python
print(f"Accuracy : {accuracy:.2%}")
print(f"Precision: {precision:.2%}")
print(f"Recall : {recall:.2%}")
```
