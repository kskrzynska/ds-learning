Zadanie 1

```python
oceny = [3.5, 4.0, 5.0, 3.0, 4.5, 5.0, 4.0, 3.5]
```

```python
def statystyki(liczby):
  """
  Zwraca tuple zawierający sumę, średnią, minimum, maksimum

  """
  suma = sum(liczby)
  srednia = round(suma / len(liczby), 2)
  minimum = min(liczby)
  maksimum = max(liczby)
  return (suma, srednia, minimum, maksimum)
```

```python
statystyki = statystyki(oceny)
```

```python
print(f"Statystyki: {statystyki}")
```

```python
temperatura = [0, 10, 20, 30, 37, 100]
```

```python
def celsius_to_fahrenheit(c):
  return c * 9 / 5 + 32
```

```python
def fahrenheit_to_celsius(f):
   return (f - 32) * 5 / 9
```

```python
def celsius_to_kelvin(c):
  return c + 273.15
```

```python
print("°C\t°F\tK")
for c in temperatura:
  f = celsius_to_fahrenheit(c)
  k = celsius_to_kelvin(c)
  print(f"{c}\t{f}\t{k}")
```

```python
def normalize_minmax(dane, metoda= "min-max", zaokraglenie=3):
  if metoda == "min-max":
    min_value = min(dane)
    max_value = max(dane)
    zakres = max_value - min_value

    znormalizowane = []
    for wartosc in dane:
        norma = (wartosc - min_value) / zakres
        znormalizowane.append(round(norma, zaokraglenie))
    return znormalizowane

```

```python
wynagrodzenia = [3500, 4200, 5800, 7200, 9500, 4800, 6100]
```

```python
wynagrodzenia_norm_minmax = normalize_minmax(wynagrodzenia)
```

```python
print(f"Oryginalne wynagrodzenia: {wynagrodzenia}")
```

```python
print(f"Normalizacja min-max: {wynagrodzenia_norm_minmax}")
```

```python
wynagrodzenia = [3500, 4200, 5800, 7200, 9500, 4800, 6100]
```

```python
def normalize_minmax(dane, metoda= "min-max", zaokraglenie=3):
  if metoda == "min-max":
    min_value = min(dane)
    max_value = max(dane)
    zakres = max_value - min_value

    return [round((x - min_value) / (max_value - min_value), zaokraglenie) for x in dane]
```

```python
print(normalize_minmax(wynagrodzenia))
```

```python
klienci = [
{"id": 1, "wiek": 25, "dochod": 4500, "wydatki": 3200},
{"id": 2, "wiek": 200, "dochod": 7800, "wydatki": 5100}, # outlier wiek
{"id": 3, "wiek": 45, "dochod": 9200, "wydatki": -500}, # outlier wydatki
{"id": 4, "wiek": 30, "dochod": 6000, "wydatki": 4000},
{"id": 5, "wiek": 50, "dochod": 11000, "wydatki": 6500},
{"id": 6, "wiek": 28, "dochod": 5000, "wydatki": 3500},
]
```

```python
def usun_outliery(klienci):
  return [klient for klient in klienci if 0 < klient["wiek"] and 120 > klient["wiek"] and 0 < klient["wydatki"]]
```

```python
print(usun_outliery(klienci))
```

```python
def normalizuj_pole(klienci, pole):
  min_value = min(klient[pole] for klient in klienci)
  max_value = max(klient[pole] for klient in klienci)

  return [ {**klient, f"{pole}_norm": round((klient[pole] - min_value) / (max_value - min_value), 3)} for klient in klienci]
```

```python
print(normalizuj_pole(klienci, "dochod"))
```
