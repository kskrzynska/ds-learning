Zadanie 10

```python
Dataset_clean = Dataset.copy()
```

```python
print("Statystyki Age przed czyszczeniem:")
print(Dataset_clean["Age"].describe())
```

```python
print("\nKształt przed czyszczeniem:", Dataset_clean.shape)
```

```python
procent_nan = Dataset_clean.isnull().mean() * 100
```

```python
kolumny_do_usuniecia = procent_nan[procent_nan >= 60].index
```

```python
Dataset_clean = Dataset_clean.drop(columns=kolumny_do_usuniecia)
```

```python
mediana_age = Dataset_clean['Age'].median()
```

```python
Dataset_clean['Age'] = Dataset_clean['Age'].fillna(mediana_age)
```

```python
dominanta_embarked = Dataset_clean['Embarked'].mode()[0]
```

```python
Dataset_clean['Embarked'] = Dataset_clean['Embarked'].fillna(dominanta_embarked)
```

```python
Dataset_clean = Dataset_clean.dropna()
```

```python
print('\nKształt po czyszczeniu:', Dataset_clean.shape)
```

```python
print('\nLiczba NaN w każdej kolumnie:')
print(Dataset_clean.isnull().sum())
```

```python
print('\nStatystyki Age po czyszczeniu:')
print(Dataset_clean['Age'].describe())
```

Zadanie 15

```python
total_passengers = 0
age_sum = 0
age_count = 0
survived_sum = 0
```

```python
for batch in pd.read_csv(url, chunksize=50):
  total_passengers += len(batch)
  age_sum += batch["Age"].sum()
  age_count += batch["Age"].count()
  survived_sum += batch["Survived"].sum()

```

```python
avg_age = age_sum / age_count
```

```python
survival_rate = survived_sum / total_passengers
```

```python
print("Wyniki batch loading:")
print("total_passengers:", total_passengers)
print("avg_age:", round(avg_age, 2))
print("survival_rate:", round(survival_rate * 100, 2), "%")
```
