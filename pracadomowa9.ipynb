```python
class Produkt:
    def __init__(self, nazwa, cena, ilosc_w_magazynie):
      self.nazwa = nazwa
      self.cena = cena
      self.ilosc_w_magazynie = ilosc_w_magazynie


    def pokaz_info(self):
      print(f"Dataset: {self.nazwa}")
      print(f"Cena: {self.cena}")
      print(f"Ilosc w magazynie: {self.ilosc_w_magazynie}")

    def sprzedaj(self, ilosc):
      self.ilosc_w_magazynie -= ilosc
      return self.ilosc_w_magazynie

    def dostawa(self, ilosc):
      self.ilosc_w_magazynie += ilosc
      return self.ilosc_w_magazynie

```

```python
produkt1 = Produkt("Samochód", 89000, 17)
produkt2 = Produkt("Rower", 1700, 36)
produkt3 = Produkt("Hulajnoga", 300, 12)
```

```python
produkt1.pokaz_info()
produkt2.pokaz_info()
produkt3.pokaz_info()
```

```python
produkt1.sprzedaj(4)
```

```python
produkt2.sprzedaj(30)
```

```python
produkt2.dostawa(12)
```

```python
produkt3.dostawa(40)
```

```python
class Kalkulator:

    def dodaj(self, a, b):
        return a + b

    def odejmij(self, a, b):
        return a - b

    def pomnoz(self, a, b):
        return a * b

    def podziel(self, a, b):
        return a / b
```

```python
kalkulator = Kalkulator()
```

```python
print(kalkulator.dodaj(19, 35))
print(kalkulator.odejmij(15, 19))
print(kalkulator.pomnoz(61, 93))
print(kalkulator.podziel(24, 8))
```

```python
class BankAccount:
  def __init__(self, wlasciciel, saldo):
    self.wlasciciel = wlasciciel
    self.saldo = saldo

  def wplac_kwota(self, kwota):
    self.saldo += kwota
    return self.saldo
  def wyplac_kwota(self, kwota):
    if kwota <= self.saldo:
      self.saldo -= kwota
      return self.saldo

  def pokaz_saldo(self):
    return self.saldo
```

```python
konto1 = BankAccount("Wlasciciel.1", 465574)
konto2 = BankAccount("Wlasciciel.2", 12234)
konto3 = BankAccount("Wlasciciel.3", 5000)
```

```python
konto1.wplac_kwota(5555)
```

```python
konto3.wplac_kwota(1)
```

```python
konto2.wyplac_kwota(12234)
```

```python
konto3.pokaz_saldo()
```

```python
class Student:
  def __init__(self, imie, numer_indeksu, oceny):
    self.imie = imie
    self.numer_indeksu = numer_indeksu
    self.oceny = []

  def dodaj_ocene(self, ocena):
    self.oceny.append(ocena)
    return self.oceny

  def srednia(self):
    return sum(self.oceny) / len(self.oceny)

  def __str__(self):
    return f"{self.imie}, numer indeksu: {self.numer_indeksu}, oceny: {self.oceny}"
```

```python
student1 = Student("Adam", 13321, ())
student2 = Student("Michał", 66532, ())
student3 = Student("Stefan", 77654, ())
```

```python
class Pracownik:
  def __init__(self, imie, stanowisko, pensja):
    self.imie = imie
    self.stanowisko = stanowisko
    self.pensja = pensja

  def podwyzka(self, procent):
    self.pensja = self.pensja * (1 + procent / 100)
    return self.pensja

  def __str__(self):
    return f"imie: {self.imie}, stanowisko: {self.stanowisko}, pensja: {self.pensja}"
```

```python
pracownik1 = Pracownik("Cezary", "księgowy", 7200)
pracownik2 = Pracownik("Szymon", "sprzątacz", 5200)
pracownik3 = Pracownik("Helena", "menadżer", 15000)
```

```python
pracownik1.podwyzka(5)
pracownik2.podwyzka(10)
pracownik3.podwyzka(50)
```

```python
print(pracownik1)
print(pracownik2)
print(pracownik3)
```

```python
class Filtrowanie_danych:
  def wykonaj(self, dane):
    return [x for x in dane if x >= 0]
```

```python
class Zaokraglanie:
  def wykonaj(self, dane):
    return [round(x) for x in dane]
```

```python
class Skalowanie:
  def __init__(self, wspolczynnik):
    self.wspolczynnik = wspolczynnik


  def wykonaj(self, dane):
    return [x * self.wspolczynnik for x in dane]
```

```python
class Sumowanie:
  def wykonaj(self, dane):
    return sum(dane)
```

```python
class Pipeline:
    def __init__(self):
        self.kroki = []

    def dodaj_krok(self, krok):
        self.kroki.append(krok)

    def uruchom(self, dane):
        wynik = dane

        for krok in self.kroki:
            wynik = krok.wykonaj(wynik)

        return wynik
```

```python
pipeline = Pipeline()
pipeline.dodaj_krok(Filtrowanie_danych())
pipeline.dodaj_krok(Zaokraglanie())
pipeline.dodaj_krok(Skalowanie(4))
pipeline.dodaj_krok(Sumowanie())
```

```python
dane = [1.7, -3.2, 5.8, -1.1, 8.3, 2.4]
```

```python
print(pipeline.uruchom(dane))
```

```python
class Akcja:
    def __init__(self, nazwa, cena_kupna, cena_aktualna):
        self.nazwa = nazwa
        self.cena_kupna = cena_kupna
        self.cena_aktualna = cena_aktualna
        self.ilosc = 0
```

```python
class Portfel:
  def __init__(self, kapital_poczatkowy):
    self.akcje = []
    self.kapital_poczatkowy = kapital_poczatkowy

  def kup_akcje(self, akcja, ilosc):
    akcja.ilosc += ilosc
    self.akcje.append(akcja)

  def sprzedaj_akcje(self, akcja, ilosc):
    akcja.ilosc -= ilosc

  def aktualizuj_ceny(self, slownik_cen):
    for akcja in self.akcje:
      akcja.cena_aktualna = slownik_cen[akcja.nazwa]

  def wartosc_portfela(self):
    suma = 0
    for akcja in self.akcje:
      suma += akcja.ilosc * akcja.cena_aktualna
    return suma

  def zysk_strata(self):
    return self.wartosc_portfela() - self.kapital_poczatkowy
```

```python
slownik_cen = {
      "Akcja1": 200,
      "Akcja2": 150,
      "Akcja3": 210,
      "Akcja4": 110,
      "Akcja5": 180
  }
```

```python
portfel = Portfel(34550)
```

```python
akcja1 = Akcja("Akcja1", 200, 150)
akcja2 = Akcja("Akcja2", 150, 250)
akcja3 = Akcja("Akcja3", 400, 350)
akcja4 = Akcja("Akcja4", 300, 200)
akcja5 = Akcja("Akcja5", 100, 150)
```

```python
portfel.kup_akcje(akcja1, 2)
```

```python
portfel.sprzedaj_akcje(akcja3, 4)
portfel.sprzedaj_akcje(akcja1, 1)
```

```python
print(portfel.wartosc_portfela())
```

```python
print(portfel.zysk_strata())
```
