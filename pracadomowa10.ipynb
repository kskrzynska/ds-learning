Zadanie 13

```python
np.random.seed(42)
data = np.concatenate([
np.random.randn(100) * 10 + 50,
np.array([150, -50, 200, -100, 180])
])
np.random.shuffle(data)
```

```python
mean = np.mean(data)
```

```python
std = np.std(data)
```

```python
data_standarized = (data - mean) / std
```

```python
outlier = np.abs(data_standarized) > 3
```

```python
outliers = data[outlier]
```

```python
print(outliers)
```

```python
usun_outliers = data[~outlier]
```

```python
mean2 = np.mean(usun_outliers)
```

```python
std2 = np.std(usun_outliers)
```

```python
data_standarized2 = (usun_outliers - mean2) / std2
```

```python
print("Statystyki przed:")
print(f"mean: {np.mean(data):.2f}")
print(f"std: {np.std(data):.2f}")
print(f"min: {np.min(data):.2f}")
print(f"max: {np.max(data):.2f}")
print(f"liczba elementów: {len(data)}")
```

```python
print("Statystyki po:")
print("mean:", np.mean(usun_outliers))
print("std:", np.std(usun_outliers))
print("min:", np.min(usun_outliers))
print("max:", np.max(usun_outliers))
print("liczba elementów:", len(usun_outliers))
```
