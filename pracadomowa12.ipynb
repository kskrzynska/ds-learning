```python
bins = [-1, 12, 18, 60, float("inf")]
```

```python
df["Kategoria Wieku"] = pd.cut( df["Age"], bins=bins, labels=labels )
```

```python
wynik = df.groupby("Kategoria Wieku", observed=True)["Survived"].mean() * 100
```

```python
print("Survival rate według kategorii wiekowych:")
print(wynik)
```

```python
print("\nWykres:")
for kategoria, procent in wynik.items():
    slupki = "█" * int(procent / 2)
    print(f"{kategoria:12} {procent:5.2f}% {slupki}")
```

```python
def make_california_housing(n=5000, seed=42):
  rng = np.random.default_rng(seed)
  latitude = rng.uniform(32.5, 41.5, n)
  longitude = rng.uniform(-123.5, -114.5, n)
  med_inc = rng.gamma(shape=2.0, scale=1.8, size=n) + 0.5
# Wiek domu
  house_age = rng.integers(1, 55, n)
# Pokoje/sypialnie/populacja
  ave_rooms = rng.normal(5.5, 1.5, n).clip(1, 15)
  ave_bedrms = (ave_rooms * rng.uniform(0.15, 0.25, n)).clip(0.5, 5)
  population = rng.integers(50, 5000, n)
  ave_occup = rng.normal(3.0, 0.8, n).clip(1, 10)
# Cena domu (w 100_000 USD) - zależna od dochodu, wieku i lokalizacji
  med_house_val = (
  med_inc * 0.6
    + (latitude - 36) * 0.15
    - (house_age / 100)
    + rng.normal(0, 0.3, n)
  ).clip(0.15, 5.0)
  return pd.DataFrame({
'MedInc': med_inc.round(4),
'HouseAge': house_age,
'AveRooms': ave_rooms.round(4),
'AveBedrms': ave_bedrms.round(4),
'Population': population,
'AveOccup': ave_occup.round(4),
'Latitude': latitude.round(4),
'Longitude': longitude.round(4),
'MedHouseVal': med_house_val.round(4)
})
```

```python
df_housing = make_california_housing()
```

```python
print(df_housing.head())
print(f"\nShape: {df_housing.shape}")
```

```python
df_housing["lat_bin"] = pd.cut(df_housing["Latitude"], 4)
```

```python
df_housing["lon_bin"] = pd.cut(df_housing["Longitude"], 4)
```

```python
df_housing["region"] = ( df_housing["lat_bin"].astype(str) + "|" + df_housing["lon_bin"].astype(str) )
```

```python
wynik = df_housing.groupby("region")["MedHouseVal"].mean()
```

```python
wynik = wynik.sort_values(ascending=False)
```

```python
print("5 regionów z najwyższą średnią ceną domu:")
print(wynik.head(5))
```

```python
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing(as_frame=True)
df = housing.frame
```

```python
df_housing["lat_bin"] = pd.cut(df_housing["Latitude"], 10)
df_housing["lon_bin"] = pd.cut(df_housing["Longitude"], 10)
```

```python
df_housing["Region"] = ( df_housing["lat_bin"].astype(str) + " | " + df_housing["lon_bin"].astype(str) )
```

```python
statystyki = df_housing.groupby("Region", observed=True).agg( Średnia_cena=("MedHouseVal", "mean"), Mediana_dochodów=("MedInc", "median"), Populacja=("Population", "sum") )
```

```python
statystyki = statystyki.sort_values("Średnia_cena", ascending=False)
```

```python
print("TOP 10 najdroższych regionów:")
print(statystyki.head(10).round(2))
```

```python
print("\nTOP 10 najtańszych regionów:")
print(statystyki.tail(10).sort_values("Średnia_cena").round(2))
```

```python
heatmap = df_housing.pivot_table( index="lat_bin", columns="lon_bin", values="MedHouseVal", aggfunc="mean", observed=False )
```

```python
q1 = heatmap.stack().quantile(0.25)
q2 = heatmap.stack().quantile(0.50)
q3 = heatmap.stack().quantile(0.75)
```

```python
print("\nTekstowa heatmapa 10 x 10:")
print("█ = najwyższe ceny")
print("■ = wysokie ceny")
print("· = niskie ceny")
print(" = najniższe ceny\n")
```

```python
for index, row in heatmap.iloc[::-1].iterrows():
  wiersz = ""
  for cena in row:
    if pd.isna(cena):
      znak = " "
    elif cena <= q1:
      znak = "·"
    elif cena <= q2:
      znak = "▪"
    elif cena <= q3:
      znak = "■"
    else: znak = "█"

    wiersz += znak + " "

  print(wiersz)
```
